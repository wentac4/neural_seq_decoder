```python
import pickle
import matplotlib.pyplot as plt
import numpy as np

def load_pickle_file(filename):
    """Loads data from a pickle file."""
    try:
        with open(filename, 'rb') as f:
            data = pickle.load(f)
        return data
    except Exception as e:
        print(f"Error loading pickle file: {e}")
        return None

pickle_filename1 = '/content/drive/MyDrive/ECEC243A/FinalProject/logs/20000_BATCH_speechBaseline4/trainingStats'
loaded_data1 = load_pickle_file(pickle_filename1)

pickle_filename2 = '/content/drive/MyDrive/ECEC243A/FinalProject/logs/20000_BATCH_ONE_POINT_FIVE_TIME_MASK_speechBaseline4/trainingStats'
loaded_data2 = load_pickle_file(pickle_filename2)

pickle_filename3 = '/content/drive/MyDrive/ECEC243A/FinalProject/logs/20000_BATCH_TWO_TIME_MASK_speechBaseline4/trainingStats'
loaded_data3 = load_pickle_file(pickle_filename3)

pickle_filename4 = '/content/drive/MyDrive/ECEC243A/FinalProject/logs/20000_BATCH_TWO_POINT_FIVE_TIME_MASK_speechBaseline4/trainingStats'
loaded_data4 = load_pickle_file(pickle_filename4)

pickle_filename5 = '/content/drive/MyDrive/ECEC243A/FinalProject/logs/20000_BATCH_THREE_TIME_MASK_speechBaseline4/trainingStats'
loaded_data5 = load_pickle_file(pickle_filename5)

test_cer1 = loaded_data1['testCER']
test_cer2 = loaded_data2['testCER']
test_cer3 = loaded_data3['testCER']
test_cer4 = loaded_data4['testCER']
test_cer5 = loaded_data5['testCER']

plt.plot(test_cer1, label='time_masking_factor = 1')
plt.plot(test_cer2, label='time_masking_factor = 1.5')
plt.plot(test_cer3, label='time_masking_factor = 2')
plt.plot(test_cer4, label='time_masking_factor = 2.5')
plt.plot(test_cer5, label='time_masking_factor = 3')

plt.xlabel('Epoch')
plt.ylabel('CER/PER')
plt.legend()
plt.show()
```
